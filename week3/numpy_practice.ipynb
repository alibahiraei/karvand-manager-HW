 تمرین دوم : پیشنهاد دهنده چند غذا همزمان

```python
import numpy as np

recipes = np.array([ 
    [15, 350, 2, 5], 
    [45, 600, 7, 10], 
    [10, 200, 0, 3], 
    [30, 450, 5, 7], 
    [60, 800, 8, 12] 
]) 
recipe_names = ["Salad", "Curry", "Toast", "Pasta", "Stew"] 
min_val=recipes.min(axis=0)
max_val=recipes.max(axis=0)
normalized = (recipes - min_val) / (max_val - min_val)
print(normalized)
print("------------------")

print("min column ",min_val)
print("max column ",max_val)
```

بخش دوم  سه مشتری با سلیقه های متفاوت داریم:

```python
users = np.array([ 
    [10, 250, 1, 4], 
    [50, 700, 8, 11], 
    [25, 400, 4, 6] 
]) 
new_normalized=(users-min_val)/(max_val-min_val)
print(new_normalized)
```

حساب کردن فاصله اقلیدوسی هر کاربر با هر غذا

```python

differences = (
    new_normalized[:, np.newaxis, :]
    - normalized[np.newaxis, :, :]
)

print(differences.shape)

distances = np.linalg.norm(differences, axis=2)
distances=np.round(distances, 3)

print(distances)
print(distances.shape)
```

برای هر کاربر نزدیکترین غذا با سلیقه ان را چاپ کنید

```python
name_users=np.array(["food","user1","user2","user3"])
distanc_food=np.vstack((recipe_names,distances))

distanc_food=np.hstack((name_users.reshape(-1,1),distanc_food))
print(distanc_food)
user1,user2,user3=np.argmin(distances,axis=1)
print("-----------------------------------")
print(f"food_user1={distanc_food[0,user1+1]}")
print(f"food_user2={distanc_food[0,user2+1]}")
print(f"food_user3={distanc_food[0,user3+1]}")

```

برای هر کار از نزدیکترین تا دورترین غذا را چاپ کنید

```python
user1,user2,user3=np.argsort(distances,axis=1)
print(f"food_user1={distanc_food[0,user1+1]}")
print(f"food_user2={distanc_food[0,user2+1]}")
print(f"food_user3={distanc_food[0,user3+1]}")
```

```python
sort_index=np.argsort(times_5k_2_new)
ranking=np.array([1,2,3,4,5,6,7,8])
sort_tim=times_5k_2_new[sort_index]
sort_bib=bib_numbers_new[sort_index]
matrix_rank=np.column_stack((ranking.reshape(-1,1),sort_bib))
matrix_rank=np.column_stack((matrix_rank,sort_tim))
print("final_rank_sort  ",sort_tim)
print("final_bib_sort ",sort_bib)
print("\n",matrix_rank)
```
